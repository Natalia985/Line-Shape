**Objective**: Calculate the natural linewidth of two transitions in neutral helium (He I):

    1. He I = First excited state to ground state
    2. He I = Second excited state to ground state
    
We are working only with neutral He, so we have two electrons. The energy levels depend on the electron configuration and momentum coupling. 


### NOTES
So He has two electrons as mentioned. So we have that: 
* 1s²: Both electrons are in the first energy level (n=1). This is the ground state.
* 1s2s: One electron remains in 1s, while the other is in the 2s orbital.
* 1s2p: One electron remains in 1s, while the other is in the 2p orbital.

s and p, they describe the type of orbital occupied by an electron, determined by the orbital angular momentum quantum number l. So, s: l=0, p: l=1, etc. For example 2p is n = 2 and l = 1.

Helium can have several states with n = 2, becuase the electrons can occupy different orbitals and have different combinations of spins. 

The equation we use for the half-width is: $$W = (2.65*10^{-20}) * (\Lambda)^2 * (A_ii + A_ff)$$

Where $\Lambda$ is the wavelength  expressed in Å.


### 1. He I - First transition to ground

In the ground state, both electrons occupy the 1s orbital (1s²). When one electron is excited to n=2, it can occupy either the 2s or 2p orbital.

The wavelength that we use is: **584.33436**. 

Also we have, 20581.287 which we can use to identify the other transition, $1s2p$ to $1s2s$. This one doesnt not reach the ground state, but it may contribute to the decay rate of the upper state. (I didn´t consider it since I wasnt sure, I left it in a documented code). 

```python
wavelength = 584.33436
A_1 = 1.7989e+09
A_ii = A_1
A_ff = 0 # Ground state 

W = (2.65e-20) * (wavelength)**2 * (A_ii + A_ff)
```

```python
# We have the half of the width of the line. 
print(W)
```

### 2. He I = Second excited state to ground state

For the second calculation we considered the transition from the energy level 3 to 1.

The wavelength that we use is: 537.02992, that correspons to the transition that was mentioned previously. 

We are using the Einstein Coefficients from transitions associated with 3 to 1 and 3 to 2. 

```python
wavelength = 537.02992
A_1 = 5.6634e+08 
A_2 = 1.299e+03
A_3 = 1.21e-01
A_ii = A_1 + A_2 + A_3 # 3 to 1 (537 wavelength)
A_4= 6.3705e+07 # 3 to 2 (667 wavelength)
A_ff = 0 # Ground state 

W = (2.65e-20) * (wavelength)**2 * (A_ii + A_4 + A_ff)
```
